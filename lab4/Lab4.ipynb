```python
import matplotlib.pyplot as plt
import numpy as np
from skimage import data, exposure
from skimage.util import img_as_float
```

```python
img = data.moon()

p3, p80 = np.percentile(img, (3, 80))
img_rescale = exposure.rescale_intensity(img, in_range=(p3, p80))

fig, axes = plt.subplots(2, 1, figsize=(10, 8))

axes[0].imshow(img_rescale, cmap="gray")
axes[0].axis("off")

axes[1].hist(img_rescale.ravel(), bins=256, range=(0, 255))
axes[1].set_xlabel("gray level values")
axes[1].set_ylabel("Number of pixels")

plt.tight_layout()
plt.show()
```

```python
img = data.moon()
img_equalized = exposure.equalize_hist(img)

fig, axes = plt.subplots(2, 1, figsize=(10, 8))

axes[0].imshow(img_equalized, cmap="gray")
axes[0].axis("off")

axes[1].hist(img_equalized.ravel(), bins=256, range=(0, 1))
axes[1].set_xlabel("gray level values")
axes[1].set_ylabel("Number of pixels")

plt.tight_layout()
plt.show()
```

```python
reference = img_as_float(data.rocket())
image = img_as_float(data.chelsea())

matched = exposure.match_histograms(
    image,
    reference,
    channel_axis=-1
)

fig, axes = plt.subplots(1, 3, figsize=(15, 5))

axes[0].imshow(reference)
axes[0].axis("off")

axes[1].imshow(image)
axes[1].axis("off")

axes[2].imshow(matched)
axes[2].axis("off")

plt.tight_layout()
plt.show()

fig, axes = plt.subplots(1, 2, figsize=(14, 5))

for channel in range(3):
    axes[0].hist(
        reference[:, :, channel].ravel(),
        bins=256,
        range=(0, 1),
        histtype="step"
    )

for channel in range(3):
    axes[1].hist(
        matched[:, :, channel].ravel(),
        bins=256,
        range=(0, 1),
        histtype="step"
    )

plt.tight_layout()
plt.show()
```
