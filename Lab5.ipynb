# Task 1

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
from skimage.data import astronaut

image = astronaut()
image_float = image.astype(np.float32) / 255.0
kernel = np.ones((7, 7), np.float32) / 49
smoothed = cv2.filter2D(image_float, -1, kernel)

fig, ax = plt.subplots(1, 2, figsize=(12, 5))
ax[0].imshow(image)
ax[0].set_title("Original Image")
ax[0].axis("off")
ax[1].imshow(np.clip(smoothed, 0, 1))
ax[1].set_title("Smoothed Image")
ax[1].axis("off")
plt.tight_layout()
plt.show()
```

# Task 2

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
from skimage.data import astronaut

image = astronaut()
filtered_5 = cv2.GaussianBlur(image, (5, 5), 0)
filtered_21 = cv2.GaussianBlur(image, (21, 21), 0)

fig, ax = plt.subplots(1, 3, figsize=(15, 5))
ax[0].imshow(image)
ax[0].set_title("Original Image")
ax[0].axis("off")
ax[1].imshow(filtered_5)
ax[1].set_title("5x5 Gaussian Filter")
ax[1].axis("off")
ax[2].imshow(filtered_21)
ax[2].set_title("21x21 Gaussian Filter")
ax[2].axis("off")
plt.tight_layout()
plt.show()
```

# Task 3

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
from skimage.data import astronaut

image = astronaut()
image_float = image.astype(np.float32) / 255.0
laplacian = cv2.Laplacian(image_float, cv2.CV_32F, ksize=3)
sharpened = np.clip(image_float - laplacian, 0, 1)

fig, ax = plt.subplots(1, 3, figsize=(15, 5))
ax[0].imshow(image_float)
ax[0].set_title("Original Image")
ax[0].axis("off")
ax[1].imshow(np.clip(laplacian, 0, 1))
ax[1].set_title("Laplacian Filter")
ax[1].axis("off")
ax[2].imshow(sharpened)
ax[2].set_title("Sharpened Image")
ax[2].axis("off")
plt.tight_layout()
plt.show()
```
