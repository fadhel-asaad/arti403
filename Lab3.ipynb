# Task 1

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

image = cv2.imread("bird.png")
image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

h, w = image.shape[:2]
resized = cv2.resize(image, (int(w * 1.5), int(h * 1.5)))

rotation_matrix = cv2.getRotationMatrix2D((w // 2, h // 2), 120, 1)
rotated = cv2.warpAffine(image, rotation_matrix, (w, h))

shear_matrix = np.float32([[1, 0.35, 0], [0, 1, 0]])
sheared = cv2.warpAffine(image, shear_matrix, (int(w * 1.35), h))

fig, ax = plt.subplots(1, 4, figsize=(18, 5))
ax[0].imshow(image)
ax[0].set_title("Original")
ax[0].axis("off")
ax[1].imshow(resized)
ax[1].set_title("Increased Size")
ax[1].axis("off")
ax[2].imshow(rotated)
ax[2].set_title("Rotated 120 Degrees")
ax[2].axis("off")
ax[3].imshow(sheared)
ax[3].set_title("Shear")
ax[3].axis("off")
plt.tight_layout()
plt.show()
```

# Task 2

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

image = cv2.imread("bird.png")
image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

negative = 255 - image
image_float = image.astype(np.float32)

c = 255 / np.log(1 + np.max(image_float))
log_image = np.uint8(np.clip(c * np.log(1 + image_float), 0, 255))

gamma = 0.6
power_image = np.uint8(255 * ((image_float / 255) ** gamma))

fig, ax = plt.subplots(1, 4, figsize=(18, 5))
ax[0].imshow(image)
ax[0].set_title("Original")
ax[0].axis("off")
ax[1].imshow(negative)
ax[1].set_title("Negative")
ax[1].axis("off")
ax[2].imshow(log_image)
ax[2].set_title("Log")
ax[2].axis("off")
ax[3].imshow(power_image)
ax[3].set_title("Power Law")
ax[3].axis("off")
plt.tight_layout()
plt.show()
```
